```python
import os, shutil, time
os.environ['HF_HUB_DISABLE_XET'] = '1'
os.environ['HF_HUB_ENABLE_HF_TRANSFER'] = '0'
from huggingface_hub import hf_hub_download, HfApi

HF_TOKEN = os.environ.get('HF_TOKEN')
REPO_ID = 'mazz92/mazz-models'
TARGET_FILENAME = 'Mazz-Technos-Draft-0.8B-Surgical.Q8_0.gguf'

print('1/3 Telechargement du checkpoint chirurgical (Weight-Space Programming)...', flush=True)
dl_p = hf_hub_download(
    repo_id='prithivMLmods/Qwen3.5-0.8B-weight-space-programming-GGUF',
    filename='Qwen3.5-0.8B-weight-space-programming.Q8_0.gguf',
    local_dir='/tmp',
    token=HF_TOKEN
)

target_p = os.path.join('/tmp', TARGET_FILENAME)
if os.path.exists(target_p) and target_p != dl_p:
    os.remove(target_p)
if dl_p != target_p:
    shutil.move(dl_p, target_p)

sz_gb = os.path.getsize(target_p) / (1024**3)
print(f'2/3 Modele opere pret : {TARGET_FILENAME} ({sz_gb:.2f} Go)', flush=True)

print(f'3/3 Upload direct vers {REPO_ID}...', flush=True)
api = HfApi(token=HF_TOKEN)
for attempt in range(1, 4):
    try:
        url = api.upload_file(
            path_or_fileobj=target_p,
            path_in_repo=TARGET_FILENAME,
            repo_id=REPO_ID,
            repo_type='model',
            commit_message=f'Add surgical model: {TARGET_FILENAME} with Weight-Space Programming (MIT CSAIL)'
        )
        print('=== SUCCES DEPLOIEMENT CHIRURGICAL ===', flush=True)
        print('URL :', url, flush=True)
        break
    except Exception as exc:
        print(f'Tentative {attempt} echouee ({exc}), retry dans 15s...', flush=True)
        time.sleep(15)
        if attempt == 3:
            raise

if os.path.exists(target_p):
    os.remove(target_p)
print('Operation chirurgicale 0.8B terminee avec succes !', flush=True)
```

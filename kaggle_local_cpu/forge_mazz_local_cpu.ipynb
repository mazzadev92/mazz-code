```python
import os, shutil, time
os.environ['HF_HUB_DISABLE_XET'] = '1'
os.environ['HF_HUB_ENABLE_HF_TRANSFER'] = '0'
from huggingface_hub import hf_hub_download, HfApi

HF_TOKEN = os.environ.get('HF_TOKEN')
REPO_ID = 'mazz92/mazz-models'
api = HfApi(token=HF_TOKEN)

print('=== [DEBUT] Mazz-Technos-Base-4B.Q4_K_M.gguf (Base Grand Public 4B CPU 4-5 tok/s) ===', flush=True)
p = hf_hub_download(repo_id='mradermacher/Huihui-Qwen3.5-4B-abliterated-GGUF', filename='Huihui-Qwen3.5-4B-abliterated.Q4_K_M.gguf', local_dir='/tmp', token=HF_TOKEN)
target_p = os.path.join('/tmp', 'Mazz-Technos-Base-4B.Q4_K_M.gguf')
if os.path.exists(target_p) and target_p != p:
    os.remove(target_p)
if p != target_p:
    shutil.move(p, target_p)
sz_gb = os.path.getsize(target_p) / (1024**3)
print(f'Telecharge: Mazz-Technos-Base-4B.Q4_K_M.gguf ({sz_gb:.2f} Go), upload vers {REPO_ID}...', flush=True)
for attempt in range(1, 4):
    try:
        url = api.upload_file(
            path_or_fileobj=target_p,
            path_in_repo='Mazz-Technos-Base-4B.Q4_K_M.gguf',
            repo_id=REPO_ID,
            repo_type='model',
            commit_message='Add sovereign model: Mazz-Technos-Base-4B.Q4_K_M.gguf (By Eren Tipirdamaz / MAZZ AENGENIUM)'
        )
        print('SUCCES: Mazz-Technos-Base-4B.Q4_K_M.gguf -> ', url, flush=True)
        break
    except Exception as exc:
        print(f'Tentative {attempt} echouee ({exc}), retry dans 15s...', flush=True)
        time.sleep(15)
        if attempt == 3:
            raise
if os.path.exists(target_p):
    os.remove(target_p)
print('=== [TERMINE] Mazz-Technos-Base-4B.Q4_K_M.gguf ===\n', flush=True)

print('=== [DEBUT] Mazz-Technos-Draft-0.8B.Q8_0.gguf (Ultra-Fast CPU Draft 0.8B) ===', flush=True)
p = hf_hub_download(repo_id='mradermacher/Qwen3.5-0.8B-abliterated-GGUF', filename='Qwen3.5-0.8B-abliterated.Q8_0.gguf', local_dir='/tmp', token=HF_TOKEN)
target_p = os.path.join('/tmp', 'Mazz-Technos-Draft-0.8B.Q8_0.gguf')
if os.path.exists(target_p) and target_p != p:
    os.remove(target_p)
if p != target_p:
    shutil.move(p, target_p)
sz_gb = os.path.getsize(target_p) / (1024**3)
print(f'Telecharge: Mazz-Technos-Draft-0.8B.Q8_0.gguf ({sz_gb:.2f} Go), upload vers {REPO_ID}...', flush=True)
for attempt in range(1, 4):
    try:
        url = api.upload_file(
            path_or_fileobj=target_p,
            path_in_repo='Mazz-Technos-Draft-0.8B.Q8_0.gguf',
            repo_id=REPO_ID,
            repo_type='model',
            commit_message='Add sovereign model: Mazz-Technos-Draft-0.8B.Q8_0.gguf (By Eren Tipirdamaz / MAZZ AENGENIUM)'
        )
        print('SUCCES: Mazz-Technos-Draft-0.8B.Q8_0.gguf -> ', url, flush=True)
        break
    except Exception as exc:
        print(f'Tentative {attempt} echouee ({exc}), retry dans 15s...', flush=True)
        time.sleep(15)
        if attempt == 3:
            raise
if os.path.exists(target_p):
    os.remove(target_p)
print('=== [TERMINE] Mazz-Technos-Draft-0.8B.Q8_0.gguf ===\n', flush=True)

```
